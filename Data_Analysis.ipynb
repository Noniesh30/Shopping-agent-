```python
import numpy as np 
import pandas as pd 
from matplotlib import pyplot as plt 
import seaborn as sns 
```

```python
PATH="Data/amazon_final.csv"
rd=pd.read_csv(PATH)
df_products=pd.DataFrame(rd)
df_products.head(5)
```

Data Analysis

```python
print("Total unique categories :",len(df_products["category_id"].unique()))
```

```python

word = "saree"
column = "title"

matches = df_products[df_products[column].astype(str).str.contains(word, case=False, na=False)]

if not matches.empty:
    pass
else:
    print("Word not found")
```

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("DataFrame Viewer")
root.geometry("900x500")
df=matches
table = ttk.Treeview(root, columns=list(df.columns), show="headings")

for col in df.columns:
    table.heading(col, text=col)
    table.column(col, width=150)

for _, row in df.iterrows():
    table.insert("", "end", values=list(row))

table.pack(fill="both", expand=True)
root.mainloop()
```
